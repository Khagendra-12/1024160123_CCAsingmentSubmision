# **Q3**

```python
import random
random.seed(1024160123)
h100 = [random.randint(100,900) for _ in range(100)]
print(h100)
```

```python
co = 0
no = []
ne = []
ce = 0
for x in h100:
  if x%2 == 0:
    co += 1
    no.append(x)
  else:
    ce += 1
    ne.append(x)
print("No of odd numbers: ", co)
print("List of odd numbers: ", no)
print("No of even numbers: ", ce)
print("List of even numbers: ", ne)
```

```python
def pr(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True

prnum = [num for num in h100 if pr(num)]

print("Prime numbers:")
print(prnum)

print("Number of prime numbers:", len(prnum))
```

```python
fr = {}

for num in h100:
    fr[num] = fr.get(num, 0) + 1

frnum = max(fr, key=fr.get)
frC = fr[frnum]

print("Most frequently occurring number:", frnum)
print("Number of occurrences:", frC)
```

# **Q4**

```python
rno = "1024160123"

digits = [int(digit) for digit in rno]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
```

```python
uset = A.union(B)

print("A union B =", uset)
```

```python
iset = A.intersection(B)

print("A intersect B =", iset)
```

```python
AminB = A.difference(B)
BminA = B.difference(A)

print("A - B =", AminB)
print("B - A =", BminA)
```

```python
Symdiff = A.symmetric_difference(B)

print("A △ B =", Symdiff)
```

```python
print("A subset B?", A.issubset(B))
print("B superset A?", B.issuperset(A))
```

```python
X = int(input("Enter a value X to remove from set A: "))

A.discard(X)

print("A after:", A)
```

```python
my_dict = {
    "name": "Khagendra Saini",
    "rno": "1024160123",
    "branch": "Computer Science and Engineering",
    "age": 20,
    "city": "Patiala"
}

print(my_dict)
```

```python
citval = my_dict.pop("city")

my_dict["location"] = citval

print(my_dict)
```

```python
my_dict["cgpa"] = 8.6

print(my_dict)
```

```python
my_dict["age"] = my_dict["age"] + 1

print(my_dict)
```

```python
dictpop = my_dict.copy()

removed_value = dictpop.pop("branch")

print("after pop():")
print(dictpop)

print("\n returned by pop():", removed_value)
```

```python
dictdel = my_dict.copy()

del dictdel["branch"]

print("after del:")
print(dictdel)
```

```python
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email does not exist")
```

```python
friend_dict = {
    "name": "Parmeet Singh",
    "roll_no": "1024160034",
    "branch": "Computers and Social Engineering",
    "age": 20,
    "city": "Patiala"
}

merged_dict = {**my_dict, **friend_dict}

print("My dictionary:")
print(my_dict)

print("\nFriend's dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)
```

```python
string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values:")
print(string_values_dict)
```
