```python
def IsPrime(n):
    if n < 2:
        return 0

    for i in range(2, n // 2 + 1):
        if n % i == 0:
            return 0

    return 1


def AddPrime(n):
    sum = 0

    for i in range(1, n + 1):
        if IsPrime(i):
            sum = sum + i

    return sum


n = int(input("Give a number: "))

print("Sum of prime numbers is -->", AddPrime(n))
```
