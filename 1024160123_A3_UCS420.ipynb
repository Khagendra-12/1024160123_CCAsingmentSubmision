# **Q6**

```python
employees_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01',
                     '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees_df = pd.DataFrame(employees_data)

employees_df

employees_df.to_csv('employees.csv', index=False)
```

```python
print("Shape:", employees_df.shape)
```

```python
average_salary = employees_df['Salary'].mean()
total_bonus = employees_df['Bonus'].sum()
youngest_age = employees_df['Age'].min()
highest_rating = employees_df['Rating'].max()

print("Average Salary:", average_salary)
print("Total Bonus:", total_bonus)
print("Youngest Employee's Age:", youngest_age)
print("Highest Performance Rating:", highest_rating)
```

```python
employees_df = employees_df.sort_values(by='Salary', ascending=False)

employees_df
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

employees_df['Performance_Category'] = employees_df['Rating'].apply(performance_category)
employees_df
```

```python
print(employees_df.isnull().sum())
```

```python
employees_df = employees_df.rename(columns={'Employee_ID': 'ID'})
employees_df
```

```python
experienced_employees = employees_df[employees_df['Years_of_Experience'] > 5]
experienced_employees
```

```python
it_employees = employees_df[employees_df['Department'] == 'IT']
it_employees
```

```python
employees_df['Tax'] = employees_df['Salary'] * 0.10
employees_df
```

```python
employees_df.to_csv('modified_employees.csv', index=False)
```
