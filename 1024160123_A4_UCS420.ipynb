```python
# Q1: Build Your Personalized Knowledge Base

import pandas as pd

fixed = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", "category": "billing"}
]

fixed += [
    {"question": "how can i change my profile name",
     "answer": "Go to Settings > Profile and edit your name.",
     "keywords": "profile name change edit", "category": "general"},
    {"question": "where can i check my payment status",
     "answer": "You can check payment status in the billing section.",
     "keywords": "payment status billing transaction", "category": "billing"}
]

df = pd.DataFrame(fixed)
print(df)
```

```python
# Q2: Generate and Score a Hypothesis

def score(q, df):
    w = q.lower().split()
    r = []

    for i, row in df.iterrows():
        k = row["keywords"].lower().split()
        s = sum(x in k for x in w)
        if s > 0:
            r.append((s, row["question"], row["answer"]))

    return sorted(r, reverse=True)

q = input("Enter your query: ")
print(score(q, df))
```

```python
# Q3: Find all questions belonging to a given category

def same_category(category_name, df):
    return df[df["category"] == category_name][["question", "category"]]

print(same_category("general", df))
```

```python
# Q4: Add a new keyword and save the updated DataFrame

i = 0
k = input("Enter a new keyword: ")
df.loc[i, "keywords"] += " " + k

df.to_csv("1024160123_faq_data.csv", index=False)
print(df)
```

```python
# Q6: Modified scoring function with tie handling

def score_tie(q, df):
    w = q.lower().split()
    r = []

    for i, row in df.iterrows():
        k = row["keywords"].lower().split()
        s = sum(x in k for x in w)
        r.append((s, row["question"], row["answer"]))

    m = max(x[0] for x in r)
    return [x for x in r if x[0] == m]

print("Tie query:")
print(score_tie("fee", df))

print("\nNon-tie query:")
print(score_tie("password", df))
```
